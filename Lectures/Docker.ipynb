### **Docker**

Docker är ett verktyg för att paketera program.

Man lägger:

- programmet
- inställningar
- bibliotek
- konfiguration

i ett paket som kallas: **Container**

#### Varför är Docker bra?

Utan Docker måste man:

- installera allt manuellt
- fixa versioner
- konfigurera systemet

Det kan bli problem: "Det fungerar på min dator men inte på din"  ---> Docker löser detta.

#### Docker Container

En container innehåller:
- appen
- bibliotek
- delar av operativsystemet
- konfiguration

Så allt fungerar likadant överallt.

- **Docker Image** := färdigt paket/recept (Sparad mall/paket)

- **Docker Container** := körande version av paketet (Körande program från imagen)

### Cloud och Datacenter:

Cloud betyder att applikationer och databaser körs på servrar i stora datacenter istället för på den egna datorn.

Vanliga cloud providers är:
- Amazon Web Services (AWS)
- Microsoft Azure
- Google Cloud

Cloud används eftersom det gör det enklare att:
- skala system
- deploya applikationer
- hantera stora mängder data
- komma åt tjänster via internet

Ett datacenter är en byggnad med tusentals servrar som kör databaser, webbappar och AI-tjänster.

### Deploy
Deploy betyder att man publicerar eller startar sin applikation på en server eller i cloudmiljö så att andra användare kan använda den.

### Kubernetes
Kubernetes används för att hantera många Docker containers samtidigt.

Kubernetes kan:
- starta containers
- stoppa containers
- automatiskt restarta krascher
- skala upp system vid hög belastning

Skillnaden är:
- Docker kör containers
- Kubernetes hanterar många containers

### Data Warehouse

Ett Data Warehouse används för strukturerad och organiserad data.

Data lagras vanligtvis i:
- tabeller
- relationer
- analysmodeller

Det används främst för:
- rapporter
- analys
- business intelligence

### Skillnad mellan Data Lake och Data Warehouse

| Data Lake | Data Warehouse |
|---|---|
| Rå och ostrukturerad data | Strukturerad data |
| Flexibel lagring | Organiserad lagring |
| Bra för AI och Big Data | Bra för analys och rapporter |

### NoSQL

NoSQL betyder “Not Only SQL”.

NoSQL-databaser använder inte traditionella relationer och tabeller på samma sätt som SQL-databaser.

De används ofta för:
- stora datamängder
- flexibel data
- realtidssystem
- hög prestanda

Exempel:
- DynamoDB
- CosmosDB
- Graph Databases


### Redis
Redis är en databas som lagrar data direkt i RAM-minnet istället för på hårddisk.

Fördel:
- extremt snabb prestanda

Nackdel:
- dyrare eftersom RAM-minne används

Redis används ofta för:
- caching
- realtidssystem
- högpresterande applikationer

### Centralisering, Lokal Teknik och Cloud

IT-branschen har genom historien gått fram och tillbaka mellan två olika sätt att arbeta: centraliserade system och lokala system.

I centraliserade system körs applikationer och data i stora datacenter och cloudmiljöer. Användarna använder sina datorer mest som klienter för att komma åt tjänster via internet. Detta är den modell som dominerar idag genom:
- cloud computing
- streamingtjänster
- webbappar
- AI-tjänster

Tidigare gick utvecklingen istället mot lokala system där användare körde program och lagrade data direkt på sina egna datorer. När persondatorer blev kraftigare kunde fler arbeta lokalt utan att vara beroende av stora centrala servrar.

Idag är vi återigen i en period där mycket centraliseras till cloud och stora datacenter. Företag som Amazon, Google och Microsoft driver enorma infrastrukturer där databaser, AI-system och webbapplikationer körs.

Samtidigt börjar man se en ny rörelse mot lokal teknik, särskilt inom AI. Fler vill kunna köra AI-modeller direkt på sina egna datorer istället för i cloud. Det finns flera orsaker till detta:
- minskat beroende av stora cloudföretag
- lägre kostnader
- bättre integritet och säkerhet
- teknisk självständighet

Teknikutveckling ofta sker i cykler. När bara stora företag har råd med avancerad teknik blir det svårare för vanliga utvecklare och mindre företag att använda och lära sig tekniken. Då uppstår ofta nya lösningar som gör tekniken mer tillgänglig igen.

Amazon Web Services (AWS) har blivit en slags standard inom cloud eftersom väldigt många företag använder deras plattform och andra leverantörer ofta anpassar sig efter AWS-modellen.

Inom databaser används PostgreSQL idag mycket brett hos många cloud providers, medan Microsoft ofta använder SQL Server och Oracle använder sina egna databassystem.

För data engineers och AI-utvecklare handlar arbetet inte bara om databaser, utan om hela ekosystemet runt data:
- cloudplattformar
- data pipelines
- data lakes
- machine learning
- AI-tjänster
- datahantering och analys

Det visar hur moderna databaser och AI-system idag är en del av mycket större distribuerade system och cloudmiljöer.

###  **Docker, Linux och Terminaler**

Docker används för att skapa isolerade miljöer där applikationer kan köras med samma versioner, bibliotek och inställningar oavsett dator eller operativsystem. Målet är att alla ska kunna köra samma projekt på samma sätt utan kompatibilitetsproblem.

En Docker Image är ett färdigt paket som innehåller applikationen och allt som behövs för att köra den. En Container är den körande versionen av imagen. För att bygga en image använder man en Dockerfile som beskriver vilka program, bibliotek och inställningar som ska ingå.

Docker containers använder nästan alltid Linux, vilket gör att kunskap om Linux och terminalen blir viktig inom Docker, cloud och Data Engineering. Många containers och servrar har inget grafiskt gränssnitt och hanteras därför via terminalkommandon.

En terminal är ett textbaserat gränssnitt där man kommunicerar med datorn genom kommandon. Ett shell, till exempel Bash, tolkar och kör dessa kommandon. Vanliga kommandon är:
- `pwd` — visar aktuell mapp
- `ls` — listar filer och mappar
- `cd` — byter mapp

För Windows-användare rekommenderas WSL eller Git Bash för att kunna arbeta i en Linux-liknande miljö.
