# CREATE LOGIN / USER / ROLE

Läraren visade:
- CREATE LOGIN
- CREATE USER
- ALTER ROLE
- GRANT EXECUTE

för att hantera användare och rättigheter i SQL Server.



# Windows Authentication vs SQL Login

| Windows Authentication | SQL Login / SQL Authentication |
|---|---|
| använder Windows-konto | använder SQL Server-konto |
| verifieras av Windows | verifieras av SQL Server |
| inget separat lösenord behövs i koden | username och password behövs |
| använder aktuell Windows-användare | använder SQL-användare som `sa` |
| säkrare i interna miljöer | mer flexibelt för externa appar |
| vanligt i företag/internt nätverk | vanligt för APIs och externa system |
| enklare att hantera | kräver hantering av lösenord |
| inget hårdkodat lösenord | lösenord kan behöva lagras |
| använder `Trusted_Connection=yes;` | använder `UID=` och `PWD=` |

---

# Windows Authentication

## Exempel

```python
connection_string = (
    "DRIVER=ODBC Driver 18 for SQL Server;"
    "SERVER=localhost;"
    "DATABASE=everyloop;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)
```
