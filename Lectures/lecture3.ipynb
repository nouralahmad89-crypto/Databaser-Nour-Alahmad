Ett bibliotek som gör databaser enklare att använda från Python. ORM gör att du kan jobba mer med Python-objekt istället för rå SQL.

### **SQLAlchemy**

Det är ett väldigt populärt Python-bibliotek.

SQLAlchemy gör flera saker

1. ORM

Arbeta med objekt istället för SQL.

2. Connection management

Hanterar kopplingar till databasen.

3. Fungerar med flera databaser

Du kan använda nästan samma Python-kod för:

- SQL Server
- PostgreSQL
- SQLite
- MySQL

Varför?

Om all logik finns i Python:

Stored Procedure =

#### **Engine och Connection**

En connection : en aktiv koppling mellan Python och databasen.

När connection finns kan vi:

- skicka SQL
- få resultat
- spara data
- uppdatera tabeller

Men först behöver vi Engine. Vad är Engine?

Engine är inte själva connection.

Engine är mer som en manager , en fabrik , en kontrollcentral

Den vet:
- vilken databas vi använder
- serveradress
- användarnamn
- lösenord
- driver

Sedan använder man: `create_engine()` för att skapa Engine-objektet.

`engine = create_engine(connection_string)` connection_string : Det är en lång text med all information för att ansluta. server , databasnamn , username , password , driver

`connection = engine.connect()` Nu öppnas den riktiga kopplingen

### SQLAlchemy — Connection & Query Basics

#### SQLAlchemy
SQLAlchemy är ett Python-bibliotek som används för att kommunicera med databaser.

Det hjälper Python att:
- ansluta till databasen
- skicka SQL queries
- hämta resultat
- arbeta med ORM eller raw SQL

---

#### Engine

Engine är ett globalt objekt som:
- håller databaskonfiguration
- vet vilken databas som används
- skapar connections

```python
engine = create_engine(url_string)
```

Engine skapas vanligtvis bara en gång.

---

#### URL String

URL/String innehåller:
- databastyp
- driver/DBAPI
- servernamn
- databasnamn
- användare/lösenord

Exempel:

```python
url_string = URL.create(
    "mssql+pyodbc",
    query={"odbc_connect": connection_string}
)
```

---

#### DBAPI / Driver

Exempel:

```text
mssql+pyodbc
```

- `mssql` = SQL Server
- `pyodbc` = driver som används för att prata med databasen

Flöde:

```text
Python
 ↓
SQLAlchemy
 ↓
pyodbc
 ↓
ODBC
 ↓
SQL Server
```

---

#### Connection

Connection är den aktiva kopplingen till databasen.

```python
with engine.connect() as conn:
```

`conn` används för:
- execute()
- queries
- läsa data
- uppdatera data

`with` används för att connection ska stängas automatiskt.

---

#### Raw SQL Query

```python
from sqlalchemy import text

query = text("""
SELECT TOP 5
    Id,
    FirstName,
    LastName
FROM Users
""")
```

`text()` gör SQL-strängen till ett SQLAlchemy SQL-object.

---

#### Execute Query

```python
result = conn.execute(query)
```

När queryn körs:
1. SQL skickas till databasen
2. databasen kör queryn
3. resultat returneras

---

#### Result Object

`execute()` returnerar först ett:

```text
Result Object / CursorResult
```

Det är inte datan direkt.

Det fungerar som en cursor mot databasen.

---

#### Fetch All Rows

```python
rows = result.fetchall()
```

`fetchall()` returnerar:
- en lista med alla rader

Exempel:

```python
[
    (1, "Nour", "Ahmad"),
    (2, "Ali", "Hassan")
]
```

---

#### Iterate Over Result

Man kan loopa direkt över result:

```python
for row in result:
    print(row.FirstName)
```

Varje `row` representerar en rad från databasen.

---

#### Access Row Data

##### Via index

```python
row[0]
```

##### Via kolumnnamn

```python
row["FirstName"]
```

##### Via attribute access

```python
row.FirstName
```

---

#### Column Names

```python
result.keys()
```

Returnerar kolumnnamnen.

Exempel:

```python
['Id', 'FirstName', 'LastName']
```
