## **Vad är normalisering?**

Normalisering betyder:

Att organisera data i databasen på ett smart sätt för att undvika problem.

Målet är att:

- minska duplicerad data
- göra uppdateringar enklare
- undvika buggar och inkonsistens
- göra databasen mer effektiv

### **1NF — First Normal Form**

Regel :  Varje kolumn måste innehålla endast ett värde per cell (atomiska värden)

- Atomiska värden :	Ett värde per cell
- Ingen lista	  : Ingen komma-separerad data
- En sak per rad  : Produkten delas upp i flera rader

### **2NF — ta bort partiella beroenden**

Regeln för 2NF .En tabell måste:
- vara i 1NF
- alla kolumner som INTE är primary key måste bero på HELA primary key

### **3NF — Third Normal Form**

Här börjar databasen bli riktigt välstrukturerad.

Regeln för 3NF : En tabell måste:

- vara i 2NF
- alla kolumner måste bero direkt på primary key
- ingen kolumn får bero på en annan icke-nyckel-kolumn
