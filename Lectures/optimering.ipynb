# Databasoptimering

Databasoptimering handlar om att göra:
- SELECT snabbare
- INSERT/UPDATE effektivare
- databasen stabil vid stora datamängder

Det viktigaste är att förstå:
- hur data lagras
- hur databasen söker
- hur queries körs internt

---

## Heap Table

En tabell utan clustered index/primär fysisk sortering.

Nya rows läggs bara till direkt.

**Fördelar**

- väldigt snabba INSERTS  
- enkelt och billigt att skriva data

**Nackdelar**

- långsamma sökningar  
- full table scan behövs ofta


**När används det?**

Bra för:
- temporär data
- logging
- staging tables
- append-only data


## Index

Ett söksystem som hjälper databasen hitta data snabbare.

Fungerar ungefär som index i en bok.


**Fördelar**

- mycket snabbare SELECT  
- snabbare WHERE/JOIN/ORDER BY  
- databasen slipper läsa hela tabellen

---

**Nackdelar**

- INSERT/UPDATE/DELETE blir långsammare  
-  index tar extra lagringsutrymme

Bra när:
- data söks ofta
- WHERE används mycket
- JOIN används mycket
- tabellen är stor.

## Clustered Index

Tabellen sorteras fysiskt efter en kolumn.

Man kan bara ha:
**ett clustered index per tabell.**


**Fördelar**

- extremt snabba sökningar  
- bra för range queries  
- bra för sortering

**Nackdelar**

- INSERT kan bli dyrare  
- om data ändras mycket → fragmentering

Bra för:
- Primary Keys
- user_id
- timestamps
- ofta sorterad data



## Non-Clustered Index


Separat indexstruktur som pekar på datan.

Själva tabellen sorteras inte.


**Fördelar**

- snabb SELECT  
- flera index kan finnas  
- flexibel


**Nackdelar**

- lite långsammare än clustered  
- extra lagring krävs

Bra för:
- email
- username
- search columns
- foreign keys


**Fragmentering**

Data och index blir utspridda i datablock/pages.

Databasen måste hoppa mellan många platser.

**Fördelar**

❌ inga 😄

---

**Nackdelar**

- långsammare queries  
- långsammare index  
- mer disk-access


Vanligt vid:
- INSERT
- UPDATE
- DELETE
- mycket dynamisk data

---

# Reindex / Defragmentering

Databasen bygger om index och organiserar om data.

**Fördelar**

- snabbare queries igen  
- bättre index-prestanda  
- mindre diskfragmentering

---

**Nackdelar**

- kostar resurser  
- kan vara tungt på stora databaser

Vanligt:
- nattjobb
- veckovis maintenance
- stora produktionssystem


## **Execution Plan**

Databasens strategi för hur queryn ska köras.

Databasen väljer:
- index
- joins
- scans
- operationer

automatiskt.

**Fördelar**

- hjälper att hitta långsamma queries  
- visar flaskhalsar  
- hjälper optimering

---

**Nackdelar***

- kan vara avancerat att läsa



När:
- querys är långsamma
- databasen känns seg
- optimering behövs


## Full Table Scan

Databasen läser hela tabellen.

**Fördelar**

- OK för små tabeller

**Nackdelar**

- mycket långsamt på stora tabeller

När:
- index saknas
- queryn inte kan använda index

---

## Views

En virtuell tabell baserad på en SELECT-query.(View = sparad SELECT-query som beter sig som en virtuell tabell)


**Fördelar**

- förenklar komplexa queries  
- kan återanvändas  
- bättre säkerhet  
- bra för rapporter



**Nackdelar**

- kan bli långsamma om de är komplexa  
- vissa views är svåra att uppdatera

---

**När används views?**

Bra för:
- dashboards
- rapporter
- BI-system
- specifika datauttag
- säkerhetskontroll

**Caching - Vad är cache?**

Cache betyder:

spara något temporärt för snabb återanvändning.
Databasen kan cacha:
- execution plans
- index-data
- query-resultat
- ofta använd data

Views kan hjälpa med:caching/förkompilering.Varför?

För viewen är:redan definierad. Databasen känner redan till:
queryn , strukturen , joins , optimeringen

## Viktigaste Praktiska Regeln

| Problem | Lösning |
|---|---|
| långsam SELECT | index |
| mycket INSERTS | heap/enkla index |
| komplex query ofta | view |
| långsam query | execution plan |
| fragmenterad data | reindex |
| stor random access-data | clustered/non-clustered index |

## Viktigaste Tankesättet

Databaser handlar inte bara om SQL.

Man måste förstå:
- hur data lagras
- hur databasen söker
- hur index fungerar
- hur queries optimeras
- hur stora datamängder påverkar prestanda
